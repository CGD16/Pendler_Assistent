```python
# [Q1] https://www.kba.de/DE/Statistik/Fahrzeuge/Fahrzeugalter/Bestand/2026/fahrzeugalter_Bestand_2026_inhalt.html
# [Q2] https://www.kba.de/DE/Statistik/Kraftverkehr/vk_methodik/2026/weiterfahrt_schlussbericht_2026_pdf.pdf?__blob=publicationFile

# auf Grund der hohen Datenmenge (über 10.8 Mio Datenpunkte = 2.5GB) werden nur die Datenunkte 
# für weitere Analysezwecke herangezogen, die bestimmte Ausschlusskriterien nicht erfüllen, wie 
# - Autos, die in DE existieren
# - Fahrzeugalter [Q1]
# - Masse-Ausreißer [Q2]
# - CO2-Emissionswerte gegen Null bei Verbrennern tbd
# - andere Vorschläge?
```

```python
# df_test = df.head(1000)
df = df[df["year"].between(2011, 2026, inclusive="both")]
df = df[df["m (kg)"].between(500, 3500, inclusive="both")]
df = df[df["Country"] == "DE"]
print("Shape: ", df.shape)
```
